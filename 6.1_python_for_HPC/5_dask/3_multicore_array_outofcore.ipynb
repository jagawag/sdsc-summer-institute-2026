# Array multicore outofcore  *(Optional deep dive)*

Out-of-core computing refers to processing data larger than memory by reading it in chunks from disk or creating chunks on the fly.

Dask supports it natively. In this example we are still processing an array with multiple threads, but the array would never fit in memory — we are creating chunks on the fly, processing them, and freeing the memory.

**You can skip this notebook** if you are comfortable with chunking from `2_multicore_array.ipynb`. The multinode notebook (`4_multinode_distributed_array.ipynb`) is the one that matters.

```python
import numpy as np
import dask.array as da
```

```python
# Try with a more computationally intensive operation
AA = da.random.normal(0,1,size=(200000,200000), chunks=(1000, 1000))
```

```python
AA.numblocks
```

```python
AA.nbytes / 1024**3
```

```python
%time AA.sum().compute()
```

```python
%time AA.sum().compute(num_workers=64)
```
